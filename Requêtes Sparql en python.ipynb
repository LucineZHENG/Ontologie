```python
from rdflib import Graph,Namespace

g = Graph()
g.parse("/content/NOUILLES.rdf", format="xml")
```

```python
#1.Quels sont les ingrédients qui composent un certain type de nouilles ?

query = """
PREFIX ns: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7#>
PREFIX nou: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7/>
SELECT ?subject ?object
WHERE {?subject ns:est_servi_avec ?object .
 }
ORDER BY ?subject

"""
qres = g.query(query)
for row in qres:
  print(f'{row.subject.split("#")[-1] if "#" in row.subject else row.subject.split("/")[-1]} est servi avec {row.object.split("#")[-1]}')
```

```python
#2.Combien de calories contiennent d’un certain type de nouilles ?
query = """
PREFIX ns: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7#>
PREFIX nou: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7/>
SELECT ?subject ?object
WHERE {
  ?subject nou:contient_des_calories ?object.
}

"""
qres = g.query(query)
for row in qres:
  print(f'{row.subject.split("#")[-1] if "#" in row.subject else row.subject.split("/")[-1]} contient des calories {row.object.split("#")[-1]}KJ')
```

```python
#5.Comment cuisiner les nouilles biangbiang ?
query = """
PREFIX ns: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7#>
PREFIX nou: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7/>

SELECT ?method ?fabrication (GROUP_CONCAT(?accompaniment; separator=", ") AS ?accompaniments)
WHERE {
  ns:NouillesBiangBiang
    ns:a_methode_de_cuisson ?method ;
    ns:a_facon_de_fabrication ?fabrication ;
    ns:est_servi_avec ?accompaniment .
}
GROUP BY ?method ?fabrication
"""

qres = g.query(query)
for row in qres:
    acc_list = [acc.split("#")[-1] for acc in str(row.accompaniments).split(", ")]
    print(f"Les nouilles biangbiang utilise la méthode de {row.method}; en utilisant des nouilles qui fait {row.fabrication}; avec des accompaniment {', '.join(acc_list)}")
```

```python
#7.Quels sont les allergènes que les nouilles sarrasin peuvent contenir ?
query = """
PREFIX ns: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7#>
PREFIX nou: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7/>
SELECT ?nouille ?allergene
WHERE {
    ?nouille ns:a_ingredient_de nou:Sarrasin .
    ?nouille ns:a_allergie_de ?allergene .
}
"""
qres = g.query(query)
for row in qres:
  print(f'Nouilles: {row.nouille.split("#")[-1] if "#" in row.nouille else row.nouille.split("/")[-1]}')
  print(f'Allergie: {row.allergene.split("#")[-1] if "#" in row.allergene else row.allergene.split("/")[-1]}')
```

```python
#8.Quel plat de nouilles est plein de protéines ?
query = """
PREFIX ns: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7#>
PREFIX nou: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7/>
PREFIX xsd: <http://www.w3.org/2001/XMLSchema#>

SELECT ?nouille ?accompaniment
WHERE {
    ?nouille ns:est_servi_avec ?accompaniment .
    ?accompaniment ns:est_riche_en_proteines "true"^^xsd:boolean .
}
"""
qres = g.query(query)
for row in qres:
  print(f'Dans {row.nouille.split("#")[-1] if "#" in row.nouille else row.nouille.split("/")[-1]} {row.accompaniment.split("#")[-1]} est plein de protéines')
```

```python
#9.Par rapport aux nouilles instantanées et nouilles sarrasin, lesquelles sont plus caloriques ?
query = """
PREFIX ns: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7#>
PREFIX nou: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7/>
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>

SELECT ?nouille ?type ?calories
WHERE {
  {
    ?nouille rdf:type nou:Nouilles_faites_à_la_machine .
    ?nouille nou:contient_des_calories ?calories .
    BIND("instantanee" AS ?type)
 }
UNION
 {
   ?nouille ns:a_ingredient_de nou:Sarrasin.
   ?nouille nou:contient_des_calories ?calories .
   BIND("sarrasin" AS ?type)
 }
}
ORDER BY DESC(?calories)
"""
qres = g.query(query)
for row in qres:
    print(f'{row.nouille.split("#")[-1] if "#" in row.nouille else row.nouille.split("/")[-1]} est nouilles {row.type.split("/")[-1]}, qui contient caloriques {row.calories}KJ')
```

```python
#13.Je veux bien dîner avec mon ami japonais, quels plats sont convenables pour nous deux ? Je suis Chinoise.
query = """
PREFIX ns: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7#>
PREFIX nou: <http://www.semanticweb.org/zhengruixing/ontologies/2025/2/untitled-ontology-7/>
PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>


SELECT ?nouille ?origine
WHERE {
  ?nouille nou:a_origine_de ?origine .
  FILTER(?origine = "Chine" || ?origine = "Japon")
}


"""
qres = g.query(query)
for row in qres:
    print(f'{row.nouille.split("#")[-1] if "#" in row.nouille else row.nouille.split("/")[-1]},{row.origine}')
```
